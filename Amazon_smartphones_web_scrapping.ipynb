## Web Scrapping Full Code**

```python
page = 1

product_price = []
product_rate = []
product_number_reviews = []
product_url = []
product_category = []
product_brand = []
product_discount = []
product_description = []
product_storage = []
product_ram = []


headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.5",
    "Accept-Encoding": "gzip, deflate, br",
    "DNT": "1",
    "Connection": "keep-alive",
    "Upgrade-Insecure-Requests": "1",
    "Sec-Fetch-Dest": "document",
    "Sec-Fetch-Mode": "navigate",
    "Sec-Fetch-Site": "none",
    "Sec-Fetch-User": "?1",
    "Cache-Control": "max-age=0",
}

session = requests.session()
session.headers.update(headers)

while page <= 180: # Max pages 
    url = f'https://www.amazon.eg/-/en/s?k=smartphone&language=en_AE&qid=1777923339&xpid=2SfWqxFFtEoc1&ref=sr_pg_{page}'
    
    response = session.get(url)

    if response.status_code != 200:
        break

    html = response.text
    soup = BeautifulSoup(html, 'html.parser')
    containers = soup.find_all('div', {'class': 'sg-col-inner'})

    if len(containers) == 0:
        break



    for container in containers:
        h2 = container.find('h2')
        if not h2:
            continue
        
        name = h2.find('span')
        if not name:
            continue
        

        # Price
        price = container.find('span', {'class': 'a-price-whole'})
        if price:
            product_price.append(price.text)
        else:
            product_price.append(None)


        # Rating
        rate = container.find('i')
        if rate:
            rate_span = rate.find('span')
            if rate_span:
                try:
                    product_rate.append(float(rate_span.text.split(' ')[0]))
                except:
                    product_rate.append(None)
            else:
                product_rate.append(None)
        else:
            product_rate.append(None)


        # Number of reviews - 0 if missing
        num_reviews = container.find('span', {'class': 'a-size-mini puis-normal-weight-text s-underline-text'})
        if num_reviews:
            try:
                product_number_reviews.append(int(num_reviews.text.replace('(', '').replace(')', '').replace(',', '')))
            except:
                product_number_reviews.append(0)
        else:
            product_number_reviews.append(0)


        # URL
        link = container.find('a', class_='a-link-normal')
        if link:
            product_url.append('https://www.amazon.eg' + link.get('href'))
        else:
            product_url.append(None)


        # Brand
        name = container.find('h2').find('span')
        product_brand.append(name.text.split()[0])


        # Discount - 0 if no discount
        discount = container.find('span', {'class': 'a-text-price'})
        if discount:
            try:
                original = float(discount.find('span', {'class': 'a-offscreen'}).text.replace('EGP', '').replace(',', '').strip())
                current_whole = container.find('span', {'class': 'a-price-whole'}).text.replace(',', '').replace('.', '').strip()
                current_fraction = container.find('span', {'class': 'a-price-fraction'}).text.strip()
                current = float(f"{current_whole}.{current_fraction}")
                discount_percent = round(((original - current) / original) * 100, 2)
                product_discount.append(discount_percent)  # Store as number, not string
            except:
                product_discount.append(0)  # 0% 
        else:
            product_discount.append(0)  # 0% discount


        # Description
        description_elem = container.find('h2', {'class': 'a-size-base-plus a-spacing-none a-color-base a-text-normal'})
        if description_elem:
            description = str(description_elem.text)
        else:
            description = None
        product_description.append(description)


        # Storage - None if not found
        if description:
            storage_match = re.search(r'(\d+\+\d+GB|\d+GB)', description)
            storage = storage_match.group(1) if storage_match else None
            product_storage.append(storage)
        else:
            product_storage.append(None)


        # RAM - None if not found
        if description:
            ram_match = re.search(r'(\d+)\+', description)
            ram = ram_match.group(1) + "GB" if ram_match else None
            product_ram.append(ram)
        else:
            product_ram.append(None)

        
    print(f'Page {page}  done')
    time.sleep(random.uniform(5, 10))
    page += 1

print(f"\nTotal products scraped: {len(product_description)}")
```

```python
# Check lengths of all lists
print(f"product_price: {len(product_price)}")
print(f"product_rate: {len(product_rate)}")
print(f"product_number_reviews: {len(product_number_reviews)}")
print(f"product_url: {len(product_url)}")
print(f"product_brand: {len(product_brand)}")
print(f"product_discount: {len(product_discount)}")
print(f"product_description: {len(product_description)}")
print(f"product_storage: {len(product_storage)}")
print(f"product_ram: {len(product_ram)}")
```
